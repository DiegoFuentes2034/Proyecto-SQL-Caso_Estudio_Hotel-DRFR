Pregunta 17. Hipótesis 1: ¿Los hijos aseguran el viaje?
Enunciado: Escribe una consulta usando CASE WHEN que clasifique a la reserva como "Con Hijos" (si children es mayor a 0 o babies es mayor a 0) o "Sin Hijos". Agrupa por esta nueva categoría y calcula la tasa de cancelación.

```python
%sql
SELECT
    CASE
        WHEN COALESCE(children, 0) > 0
          OR COALESCE(babies, 0) > 0
        THEN 'Con Hijos'
        ELSE 'Sin Hijos'
    END AS categoria_familiar,
    COUNT(*) AS total_reservas,
    ROUND(AVG(is_canceled) * 100, 2) AS tasa_cancelacion
FROM cursodataacademy.proyecto_hotel.hotel_bookings
GROUP BY
    CASE
        WHEN COALESCE(children, 0) > 0
          OR COALESCE(babies, 0) > 0
        THEN 'Con Hijos'
        ELSE 'Sin Hijos'
    END;
```

conclusion: La consulta permite comparar las tasas de cancelación entre las reservas con hijos y sin hijos, identificando qué segmento presenta mayor riesgo de cancelación.

Conclusión analítica: Las reservas sin hijos presentan una mayor tasa de cancelación (37.22 %) frente a las reservas con hijos (34.92 %). Esto indica que los huéspedes que viajan sin hijos tienen una mayor tendencia a cancelar sus reservas.

Pregunta 18. Hipótesis 2: El impacto de las solicitudes especiales
Enunciado: Agrupa por la columna total_of_special_requests y calcula la tasa de cancelación (AVG(is_canceled)) para cada cantidad de solicitudes. Ordena por la cantidad de solicitudes de forma ascendente.

```python
%sql
SELECT
    total_of_special_requests,
    COUNT(*) AS total_reservas,
    ROUND(AVG(is_canceled) * 100, 2) AS tasa_cancelacion
FROM cursodataacademy.proyecto_hotel.hotel_bookings
GROUP BY total_of_special_requests
ORDER BY total_of_special_requests ASC;
```

conclusion: La consulta permite evaluar la relación entre las solicitudes especiales y las cancelaciones, identificando si los clientes que realizan más solicitudes presentan una mayor o menor probabilidad de cancelar.

Conclusión analítica: Las reservas sin solicitudes especiales presentan la mayor tasa de cancelación (47.72 %). Asimismo, se observa que, en general, a mayor cantidad de solicitudes especiales, menor es la tasa de cancelación, lo que sugiere un mayor compromiso del huésped.

Pregunta 19. Hipótesis 3: ¿Cambiar la reserva es señal de compromiso?
Enunciado: Utiliza CASE WHEN para segmentar la columna booking_changes en tres grupos: "Sin Cambios" (0), "1 a 3 cambios" y "Más de 3 cambios". Calcula la tasa de cancelación para cada grupo.

```python
%sql
SELECT
    CASE
        WHEN booking_changes = 0 THEN 'Sin Cambios'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
        ELSE 'Más de 3 cambios'
    END AS categoria_cambios,
    COUNT(*) AS total_reservas,
    ROUND(AVG(is_canceled) * 100, 2) AS tasa_cancelacion
FROM cursodataacademy.proyecto_hotel.hotel_bookings
GROUP BY
    CASE
        WHEN booking_changes = 0 THEN 'Sin Cambios'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
        ELSE 'Más de 3 cambios'
    END
ORDER BY tasa_cancelacion ASC;
```

conclusion: La consulta permite identificar qué grupo de clientes presenta menor tasa de cancelación según la cantidad de modificaciones realizadas en sus reservas, facilitando la evaluación del compromiso de los huéspedes.

Conclusión analítica: Las reservas con 1 a 3 modificaciones presentan la menor tasa de cancelación (15.59 %), mientras que aquellas sin modificaciones alcanzan el 40.85 %. Esto sugiere que realizar cambios en la reserva puede estar relacionado con un mayor compromiso del huésped.

Pregunta 20. Hipótesis 4: El factor precio (adr)
Enunciado: Agrupa las reservas por tipo de hotel (hotel). Para cada tipo, muestra la cantidad total de cancelaciones y el precio promedio (adr) solamente de las reservas canceladas.

```python
%sql
SELECT
    hotel,
    COUNT(*) AS total_cancelaciones,
    ROUND(AVG(adr), 2) AS precio_promedio_cancelaciones
FROM cursodataacademy.proyecto_hotel.hotel_bookings
WHERE is_canceled = 1
GROUP BY hotel
ORDER BY precio_promedio_cancelaciones DESC;
```

conclusion: La consulta permite comparar la cantidad de cancelaciones y el precio promedio de las reservas canceladas por tipo de hotel, identificando dónde se concentra el mayor valor económico de las cancelaciones.

Conclusión analítica: El Resort Hotel presenta el precio promedio más alto en las reservas canceladas (USD 105.79), frente al City Hotel (USD 104.69). Además, el City Hotel concentra la mayor cantidad de cancelaciones, con 33,102 registros.
